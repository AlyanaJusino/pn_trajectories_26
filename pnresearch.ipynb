```python
import numpy as np 
```

```python
N = 100
t = np.linspace(0,1,N)
y = np.zeros(N)

y0 = 10
y[0] = y0
```

```python
def f(y,t):
    return 0.1
```

```python
def solve(y,t,f): 
    for n in range(1, len(t)):
        y[n]= y[n-1] + (t[n] - t[n-1]) * f(y[n-1], t[n-1]);

    return None
```

```python
N = 100
t = np.linspace(0,1,N)

def setup_and_solve(y0, f,t):
    y = np.zeros(len(t))
    y[0] = y0; 

    solve(y,t,f)
    return y
```

```python
y = setup_and_solve(t=t, y0=0.01, f=f)
```

2nd order 

```python
def f(y,t):
    return 0.1 + y * 0.05
```

```python
def solve(y,t,f): 
    for n in range(1, len(t)):
        y[:,n]= y[:,n-1] + (t[n] - t[n-1]) * f(y[:,n-1], t[n-1]);

    return None
```

```python
def setup_and_solve(y0, f,t):
    y = np.zeros((len(y0),len(t)))
    y[:,0] = y0; 

    solve(y,t,f)
    return y
```

```python
y = setup_and_solve(t=t, y0=np.array([0.0,0.1]), f=f)
```
