# 1st Order/1D

```python
import numpy as np 
```

```python
N = 100
t = np.linspace(0,1,N)
y = np.zeros(N)

y0 = 10
y[0] = y0
```

```python
def f(y,t):
    return 0.1
```

```python
def solve(y,t,f): 
    for n in range(1, len(t)):
        y[n]= y[n-1] + (t[n] - t[n-1]) * f(y[n-1], t[n-1]);

    return None
```

```python
N = 100
t = np.linspace(0,1,N)

def setup_and_solve(y0, f,t):
    y = np.zeros(len(t))
    y[0] = y0; 

    solve(y,t,f)
    return y
```

```python
y = setup_and_solve(t=t, y0=0.01, f=f)
```

# 2nd order 

```python
def f(y,t): 
    return 0.1 + y * 0.05
```

```python
def solve(y,t,f): 
    for n in range(1, len(t)):
        y[:,n]= y[:,n-1] + (t[n] - t[n-1]) * f(y[:,n-1], t[n-1]);

    return None
```

```python
def setup_and_solve(y0, f,t):
    y = np.zeros((len(y0),len(t)))
    y[:,0] = y0; 

    solve(y,t,f)
    return y
```

```python
y = setup_and_solve(t=t, y0=np.array([0.0,0.1]), f=f)
```

# ASSIGNMENT PT 2
- create a function for two point masses on a spring, for: 
  - x** and y**
    

```python
#initial conditions given 
L_0 = 5
k = 1
m = 1
```

```python
#start of our spring equation  for two-bodies
def spring_eq_twomasses(y,t): 
    #this time we are unpacking the vectors for each body
    x1 = y[0] #position of mass1 
    v1 = y[1] #velocity of mass 1
    x2 = y[2] #position mass 2
    v2 = y[3] #velocity of mass 2

    #the stretch eq thingy
    stretch = (x2 - x1) - L_0
    
    #[velocity1, a**1, velocity 2, a**2] 
    return np.array([v1, (k*stretch)/m, v2, (-k*stretch)/m])
```

```python
N = 5000 #how many steps or points it takes

y0 = np.array([0.5, 1, 0.5, 0]) #given initial conditions

t = np.linspace(0,20,N) #start at 0 seconds and end at 20 seconds 

y = setup_and_solve(y0=y0, t=t, f=spring_eq_twomasses)
```
