```python
import numpy as np 
```

```python
N = 100
t = np.linspace(0,1,N)
y = np.zeros(N)

y0 = 10
y[0] = y0
```

```python
def f(y,t):
    return 0.1
```

```python
def solve(y,t,f): 
    for n in range(1, len(t)):
        y[n]= y[n-1] + (t[n] - t[n-1]) * f(y[n-1], t[n-1]);

    return None
```

```python
N = 100
t = np.linspace(0,1,N)

def setup_and_solve(y0, f,t):
    y = np.zeros(len(t))
    y[0] = y0; 

    solve(y,t,f)
    return y
```

```python
y = setup_and_solve(t=t, y0=0.01, f=f)
```

# 2nd order 

```python
def f(y,t): 
    return 0.1 + y * 0.05
```

```python
def solve(y,t,f): 
    for n in range(1, len(t)):
        y[:,n]= y[:,n-1] + (t[n] - t[n-1]) * f(y[:,n-1], t[n-1]);

    return None
```

```python
def setup_and_solve(y0, f,t):
    y = np.zeros((len(y0),len(t)))
    y[:,0] = y0; 

    solve(y,t,f)
    return y
```

```python
y = setup_and_solve(t=t, y0=np.array([0.0,0.1]), f=f)
```

```python
y
```

# ASSIGNMENT
- CODE FOR A SPRING X** = -X
- CURVE OF POSITION AS A FUNCTION OF TIME
- CURVE OF VELOCITY AS A FUNCTION OF TIME 

```python
#start of our spring equation 
def spring_eq(y,t): 
    return np.array([y[1], -y[0]]) #this is the vectors of the position and velocity
    #y(0) is position
    #y(1) is velocity 
    
```

```python
N = 5000 #how many steps or points it takes
y0 = np.array([1.0, 0.0]) #initial, position 1 is 

t = np.linspace(0,20,N) #start at 0 seconds and end at 20 seconds 

y = setup_and_solve(y0=y0, t=t, f=spring_eq) #change f to f = spring_eq to call 
#back to the function
```

```python
plt.plot(t, y[0])

plt.xlabel("Time")
plt.ylabel("Position")
plt.title("Position v. Time")

plt.grid() #because grid lines look nice sometimes
plt.show()
```

```python
plt.plot(t, y[1])
plt.xlabel("Time")
plt.ylabel("Velocity")
plt.title("Velocity v. Time")

plt.grid() #because grid lines look nice sometimes
plt.show()
```

```python
fig, ax = plt.subplots() #making the overlay

#overlaying the plots 
ax.plot(t, y[1], label='Velocity', color='red') 
ax.plot(t,y[0], label='Position', color='blue')
ax.set_title('Overlay Plots')
ax.legend()
ax.grid(True)
```

### redoing it with the damping (?) force
- so x** = -x*-0.01v

```python
#start of our spring equation (damp) 
def spring_eq(y,t): 
    return np.array([y[1], -y[0] - 0.1 * y[1]]) #this is the vectors of the position and velocity
    #y(0) is position
    #y(1) is velocity 
    # now including the damping force of the equation , play with the -0.01, making it -1.00 cancels it 
    
```

```python
N = 1000 #how many steps or points it takes

y0 = np.array([1.0, 0.0]) #initial, position 1 is 

t = np.linspace(0,20,N) #start at 0 seconds and end at 20 seconds 

y = setup_and_solve(y0=y0, t=t, f=spring_eq) #change f to f = spring_eq to call 
#back to the function
```

```python
plt.plot(t, y[0], label="Position")
plt.plot(t, y[1], label="Velocity")
plt.xlabel("Time")
plt.ylabel("Position / Velocity")
plt.title("Damped Spring")
plt.legend()
plt.grid(True)
plt.show()
```
